# 1. Build and inspect

Core owns the physical building. Data supplies constructions and environmental factors. EPSM connects them to energy calculations and LCA. This example uses a clearly selected teaching building with Data’s existing construction catalogue. Its dimensions and use are assumptions, not observations. No energy engine runs here.

```python
from renovation_case import make_case

from epsm import Building

draft = Building()
print("An empty building still needs:", draft.readiness("quantities"))
case = make_case()
baseline = case["baseline"]
print(baseline)
print("Chosen teaching preset:", baseline.preset)
print("Ready to count materials:", baseline.readiness("quantities"))
print("Load reference area (m²):", baseline.effective_load_reference_area_m2)
print("Thermal bridge uplift:", baseline.thermal_bridge_uplift_fraction)
print("Ground choice:", baseline.foundation.method)
```

The shared helper explicitly selects Core’s illustrative reference preset, assigns Data’s existing multi-family construction set, and states district-heating efficiency and fan power. Inspect the helper before using the case for a study. Material quantities preserve their actual bases; unknown window mass stays unknown.

```python
quantities = baseline.quantities()
for row in quantities.by_material():
    print(row)

settings = case["settings"]
for material_id, factors in settings.material_factors.items():
    print(
        material_id,
        {
            stage: (factor.value, factor.unit, factor.source)
            for stage, factor in factors.items()
        },
    )
```

A source record may describe only part of a building. Missing physical information remains missing. This small record supplies a footprint and storey heights; it does not supply services, operation or a ground model.

```python
record = {
    "building_id": "record-example",
    "storeys": 1,
    "facts": {"storey_height_m": 3},
    "geometry": {
        "type": "Polygon",
        "coordinates": [[[0, 0], [10, 0], [10, 8], [0, 8], [0, 0]]],
    },
}
observed = Building.from_record(record, construction_set=baseline.construction_set)
print(observed.readiness())
print(observed.compare())
assert observed.hvac is None
assert observed.foundation is None
```
